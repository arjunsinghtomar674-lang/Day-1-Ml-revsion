```python
street_freq = df['street'].value_counts()
df['street'] = df['street'].map(street_freq)
df['street'] = df['street'].map(street_freq).fillna(1)
```

```python
street_freq = df['statezip'].value_counts()

df['statezip'] = df['statezip'].map(street_freq)
df['statezip'] = df['statezip'].map(street_freq).fillna(0)
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
oe = OneHotEncoder(sparse_output = False ,handle_unknown = 'ignore')
```

```python
df = pd.get_dummies(df, columns=['city'], dtype=int)
```

```python
df.head()
```

```python
df['date'] = pd.to_datetime(df['date'])
```

```python
df['years'] = df['date'].dt.year
df['months'] = df['date'].dt.month
df['day'] = df['date'].dt.day
df['dayofweek'] = df['date'].dt.dayofweek
```

```python
df.drop(columns = ['date'],axis = 1,inplace=True)
```

```python
df.head()
```

```python
df = df.drop(columns = ['country'],axis = 1)
```

```python
X = df.drop(columns = ['price'],axis=1)
y = df['price']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(
    X,y,test_size = 0.33,random_state=42
)
```

```python
from sklearn.linear_model import Ridge
```

```python
rd = Ridge()
```

```python
rd.fit(X_train,y_train)
```

```python
y_pred = rd.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
mean_absolute_error(y_test,y_pred)
```

```python
mean_squared_error(y_test,y_pred)
```

```python
r2_score(y_test,y_pred)
```

```python
new_house = X_test.iloc[[0]]

prediction = rd.predict(new_house)
```
