# Cross Validation Score

# 1. K-Fold

```python
import pandas as pd

from sklearn.model_selection import KFold, cross_val_score
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score
```

```python
df = pd.read_csv(r'C:\\Users\\Arjun Singh Tomar\\Downloads\\clean_synthetic_regression.csv')
```

```python
df.head()
```

```python
X = df.drop("Salary_LPA", axis=1)
y = df["Salary_LPA"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)
```

```python
kf = KFold(
    n_splits = 5,
    shuffle = True,
    random_state = 42
)
```

```python
scores = cross_val_score(
    model,
    X,
    y,
    cv = kf,
    scoring = 'r2'
)
```

```python
print("Fold Scores:")
print(scores)

print("\nMean CV Score:")
print(scores.mean())

print("\nStandard Deviation:")
print(scores.std())
```
