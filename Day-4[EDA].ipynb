```python
# Outlier ana;ysis 
```

```python
sns.boxplot(x="Fare",data=df)
```

```python
Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
```

```python
outliers = df[
    (df["Fare"] < lower) |
    (df["Fare"] > upper)
]
```

# Using loop for graphs

```python
# Numerical coluns only
```

```python
num_cols = df.select_dtypes(include="number").columns
```

```python
print(num_cols)
```

```python
num_cols = df.select_dtypes(include="number").columns

for col in num_cols:
    sns.histplot(data=df, x=col, kde=True)
    plt.title(f"Distribution of {col}")
    plt.show()
```

```python
# Categorical column only
```

```python
cat_cols = df.select_dtypes(include="object").columns
```

```python
print(cat_cols)
```

```python
for col in cat_cols:
    sns.countplot(x=col,data=df)
    plt.title(f"Distribution of {col}")
    plt.show()
```
