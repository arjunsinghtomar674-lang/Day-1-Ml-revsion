# Stratified K-Fold

```python
import pandas as pd

from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import r2_score
```

```python
df = pd.read_csv(r'C:\\Users\\Arjun Singh Tomar\\Downloads\\classification_practice.csv')
```

```python
df.head()
```

```python
X = df.drop("Purchased", axis=1)
y = df["Purchased"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)
```

```python
skf = StratifiedKFold(
    n_splits = 5,
    shuffle = True,
    random_state = 42
)
```

```python
score = cross_val_score(
    model,
    X,
    y,
    cv = skf,
    scoring="accuracy"
)
```

```python
print(score)
```

```python
print(score.mean())
```
